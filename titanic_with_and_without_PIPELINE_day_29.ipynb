```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder
from sklearn.preprocessing import OrdinalEncoder
from sklearn.tree import DecisionTreeClassifier
```

```python
df=pd.read_csv('titanic.csv')
```

```python
df.head(

)
```

```python
df.drop(columns=['PassengerId','Name','Ticket','Cabin'],inplace=True)
```

```python
df
```

```python
x_train,x_test,y_train,y_test=train_test_split(df.drop(columns=['Survived']),
                                               df['Survived'],
                                               test_size=0.2,
                                               random_state=42)
```

```python
#Applyong Simple IN=mputeer
```

```python
si_age = SimpleImputer()
si_embarked = SimpleImputer(strategy='most_frequent')

x_train_age=si_age.fit_transform(x_train[['Age']])
x_train_embarked=si_embarked.fit_transform(x_train[['Embarked']])

x_test_age=si_age.fit_transform(x_test[['Age']])
x_test_embarked=si_embarked.fit_transform(x_test[['Embarked']])
```

```python
#ONe hot encodeing the Sex and Emabrked4

ohe_sex=OneHotEncoder(handle_unknown='ignore')
ohe_embarked=OneHotEncoder(handle_unknown='ignore')

x_train_sex= ohe_sex.fit_transform(x_train[['Sex']])
x_train_embarked=ohe_embarked.fit_transform(x_train_embarked)

x_test_sex= ohe_sex.fit_transform(x_test[['Sex']])
x_test_embarked=ohe_embarked.fit_transform(x_test_embarked)
```

```python
x_train.head(2)
```

```python
x_train_rem = x_train.drop(columns=['Sex','Age','Embarked'])
x_test_rem = x_test.drop(columns=['Sex','Age','Embarked'])
```

```python
from scipy.sparse import hstack

print("Shape of x_train_rem.values:", x_train_rem.values.shape)
print("Shape of x_train_age:", x_train_age.shape)
print("Shape of x_train_sex:", x_train_sex.shape)
print("Shape of x_train_embarked:", x_train_embarked.shape)
print("Shape of x_test_rem.values:", x_test_rem.values.shape)
print("Shape of x_test_age:", x_test_age.shape)
print("Shape of x_test_sex:", x_test_sex.shape)
print("Shape of x_test_embarked:", x_test_embarked.shape)

x_train_transformed = hstack((x_train_rem.values, x_train_age, x_train_sex, x_train_embarked))
x_test_transformed = hstack((x_test_rem.values, x_test_age, x_test_sex, x_test_embarked))
```

```python
x_test_transformed.shape
```

```python
clf=DecisionTreeClassifier()
clf.fit(x_train_transformed,y_train)
```

```python
y_pred=clf.predict(x_test_transformed)
y_pred
```

```python
from sklearn.metrics import accuracy_score
accuracy_score(y_test,y_pred)
```
